```python
# Import required libraries
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
# Import MySQL connector
import mysql.connector
print("MySQL connector imported successfully!")
```

```python
# Connect to the HR Analytics MySQL database

connection = mysql.connector.connect(
    host="localhost",
    user="root",
    password="1612",
    database="enterprise_hr_analytics",
    port=3306
)

if connection.is_connected():
    print("MySQL connection successful!")
    print("Connected database: enterprise_hr_analytics")
```

```python
# Load employees table

employees_df = pd.read_sql(
    "select * from employees;",
    connection
)

print("Employees loaded:", employees_df.shape)
```

```python
# Load employees table

performance_df = pd.read_sql(
    "select * from performance_ratings;",
    connection
)

print("Performance ratings loaded:", performance_df.shape)
```

```python
# Record original dataset information

print("ORIGINAL DATASET SUMMARY")

print("\nEmployees:")
print("Rows:", employees_df.shape[0])
print("Columns:", employees_df.shape[1])
print("Duplicates:", employees_df.duplicated().sum())

print("\nPerformance Ratings:")
print("Rows:", performance_df.shape[0])
print("Columns:", performance_df.shape[1])
print("Duplicates:", performance_df.duplicated().sum())
```

```python
# Standardize column names

employees_df.columns =(
    employees_df.columns.str.strip().str.replace(" ","_")
)
performance_df.columns =(
    performance_df.columns.str.strip().str.replace(" ","_")
)

print("Empoyees columns:", employees_df.columns.to_list())
print("\nPerformance Ratings columns:", performance_df.columns.to_list())
```

```python
# Check missing values in employees table
print("EMPLOYEES MISSING VALUES")
employees_missing = employees_df.isnull().sum()
print(employees_missing[employees_missing > 0])
```

```python
# Check missing values in performance ratings table
print("PERFORMANCE RATINGS MISSING VALUES ")
performance_missing = performance_df.isnull().sum()
print(performance_missing[performance_missing > 0])
```

```python
# Analyze missing values

print("EMPLOYEES TABLE")

employees_missing = pd.DataFrame({
    "Missing_Count": employees_df.isnull().sum(),
    "Missing_Percentage": (
        employees_df.isnull().sum() / len(employees_df) * 100
    ).round(2)
})

print(employees_missing[employees_missing["Missing_Count"] > 0])


print("\nPERFORMANCE RATINGS TABLE")

performance_missing = pd.DataFrame({
    "Missing_Count": performance_df.isnull().sum(),
    "Missing_Percentage": (
        performance_df.isnull().sum() / len(performance_df) * 100
    ).round(2)
})

print(performance_missing[performance_missing["Missing_Count"] > 0])
```

```python
# Check duplicate records

print("DUPLICATE CHECK")

print("Employees duplicate rows:",
      employees_df.duplicated().sum())

print("Performance ratings duplicate rows:",
      performance_df.duplicated().sum())
```

```python
# Check current data types

print(" EMPLOYEES DATA TYPES")
print(employees_df.dtypes)

print("\nPERFORMANCE RATINGS DATA TYPES")
print(performance_df.dtypes)
```

```python
# Convert HireDate to datetime

employees_df["HireDate"] = pd.to_datetime(
    employees_df["HireDate"],
    errors="coerce"
)

print("HireDate data type:",
      employees_df["HireDate"].dtype)
```

```python
# Convert ReviewDate to datetime

performance_df["ReviewDate"] = pd.to_datetime(
    performance_df["ReviewDate"],
    format="%m/%d/%Y",
    errors="coerce"
)

print("ReviewDate data type:",
      performance_df["ReviewDate"].dtype)
```

```python
# Check invalid dates after conversion

print("Invalid HireDate values:",
      employees_df["HireDate"].isna().sum())

print("Invalid ReviewDate values:",
      performance_df["ReviewDate"].isna().sum())
```

```python
# Preview cleaned date columns

print("Employee Hire Dates:")
display(
    employees_df[["EmployeeID", "HireDate"]].head(10)
)

print("\nPerformance Review Dates:")
display(
    performance_df[
        ["PerformanceID", "EmployeeID", "ReviewDate"]
    ].head(10)
)
```

```python
# Validate Age

print("Age minimum:", employees_df["Age"].min())
print("Age maximum:", employees_df["Age"].max())

print("\nInvalid Age records:")
display(
    employees_df[
        (employees_df["Age"] < 18) |
        (employees_df["Age"] > 70)
    ][["EmployeeID", "Age"]]
)
```

```python
# Validate YearsAtCompany

print("YearsAtCompany minimum:",
      employees_df["YearsAtCompany"].min())

print("YearsAtCompany maximum:",
      employees_df["YearsAtCompany"].max())

print("\nInvalid YearsAtCompany records:")

display(
    employees_df[
        (employees_df["YearsAtCompany"] < 0) |
        (employees_df["YearsAtCompany"] > employees_df["Age"])
    ][["EmployeeID", "Age", "YearsAtCompany"]]
)
```

```python
# Validate satisfaction ratings

satisfaction_columns = [
    column for column in employees_df.columns
    if "Satisfaction" in column
]

print("Satisfaction columns found:")
print(satisfaction_columns)

print("\nSATISFACTION RATING VALIDATION")

for column in satisfaction_columns:
    print(
        column,
        "| Min:", employees_df[column].min(),
        "| Max:", employees_df[column].max()
    )

    # Check values outside the expected 1–5 range
    invalid_values = employees_df[
        (employees_df[column] < 1) |
        (employees_df[column] > 5)
    ]

    print("Invalid records:", len(invalid_values))
    print()
```

```python
# Validate performance ratings

rating_columns = [
    "SelfRating",
    "ManagerRating"
]

print("PERFORMANCE RATING VALIDATION")

for column in rating_columns:
    if column in performance_df.columns:
        print(
            column,
            "| Min:", performance_df[column].min(),
            "| Max:", performance_df[column].max()
        )

        # Check values outside expected 1–5 range
        invalid_values = performance_df[
            (performance_df[column] < 1) |
            (performance_df[column] > 5)
        ]

        print("Invalid records:", len(invalid_values))
    else:
        print("Column not found:", column)
```

```python
# Check categorical values

categorical_columns = [
    "Gender",
    "BusinessTravel",
    "Department",
    "State",
    "Ethnicity",
    "JobRole",
    "MaritalStatus"
]

print("CATEGORICAL VALUE CHECK")

for column in categorical_columns:
    if column in employees_df.columns:
        print(f"\n{column}:")
        print(employees_df[column].value_counts(dropna=False))
    else:
        print(f"\nColumn not found: {column}")
```

```python
# Handling Missing Values
# Check missing values before cleaning

print("EMPLOYEES")
print(employees_df.isnull().sum()[employees_df.isnull().sum() > 0])

print("\nPERFORMANCE RATINGS")
print(performance_df.isnull().sum()[performance_df.isnull().sum() > 0])

# Fill missing numerical values with median
numeric_employee_columns = employees_df.select_dtypes(
    include="number"
).columns
for column in numeric_employee_columns:
    if employees_df[column].isnull().sum() > 0:
        employees_df[column] = employees_df[column].fillna(
            employees_df[column].median()
        )
numeric_performance_columns = performance_df.select_dtypes(
    include="number"
).columns

for column in numeric_performance_columns:
    if performance_df[column].isnull().sum() > 0:
        performance_df[column] = performance_df[column].fillna(
            performance_df[column].median()
        )
print("\nMissing numerical values handled.")

# Fill missing categorical values
categorical_employee_columns = employees_df.select_dtypes(
    include="object"
).columns
for column in categorical_employee_columns:
    employees_df[column] = employees_df[column].fillna("Unknown")
categorical_performance_columns = performance_df.select_dtypes(
    include="object"
).columns
for column in categorical_performance_columns:
    performance_df[column] = performance_df[column].fillna("Unknown")
print("\nMissing categorical values handled.")

```

```python
# Remove duplicate records

employee_duplicates = employees_df.duplicated().sum()
performance_duplicates = performance_df.duplicated().sum()

print("Employee duplicate rows:", employee_duplicates)
print("Performance duplicate rows:", performance_duplicates)

if employee_duplicates > 0:
    employees_df = employees_df.drop_duplicates()

if performance_duplicates > 0:
    performance_df = performance_df.drop_duplicates()

print("\nDuplicates handled successfully.")
print("Employees rows:", len(employees_df))
print("Performance rows:", len(performance_df))
```

```python
# Verify text columns after cleaning

for column in text_columns:
    spaces_found = (
        employees_df[column]
        .astype("string")
        .str.strip()
        .ne(employees_df[column].astype("string"))
        .sum()
    )

    if spaces_found > 0:
        print(f"{column}: {spaces_found} spaces still found")

print("Text cleaning verification completed.")
```

```python
# Check for leading or trailing spaces in text columns

text_columns = employees_df.select_dtypes(include="object").columns
for column in text_columns:
    if employees_df[column].astype(str).str.strip().ne(
        employees_df[column].astype(str)
    ).any():
        print("Spaces found in:", column)
```

```python
# Remove leading and trailing spaces

for column in text_columns:
    employees_df[column] = (
        employees_df[column]
        .astype("string")
        .str.strip()
    )

for column in performance_df.select_dtypes(
    include="object"
).columns:
    performance_df[column] = (
        performance_df[column]
        .astype("string")
        .str.strip()
    )

print("Text formatting standardized.")
```
