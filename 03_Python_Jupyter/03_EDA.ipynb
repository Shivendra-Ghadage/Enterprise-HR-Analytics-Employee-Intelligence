```python
# Compare Average Self and Manager Ratings

average_ratings = performance_df[
    ["SelfRating", "ManagerRating"]
].mean().round(2)
print("Average Ratings:")
print(average_ratings)
```

```python
# Self Rating vs Manager Rating

plt.figure(figsize=(8, 5))
average_ratings.plot(kind="bar")
plt.title("Average Self Rating vs Manager Rating")
plt.xlabel("Rating Type")
plt.ylabel("Average Rating")
plt.xticks(rotation=0)
plt.show()
```

```python
# Calculate Rating Difference

performance_df["RatingDifference"] = (
    performance_df["SelfRating"] -
    performance_df["ManagerRating"]
)
print("Rating Difference Statistics:")
print(performance_df["RatingDifference"].describe())
```

```python
# Rating Agreement Analysis

rating_agreement = performance_df["RatingDifference"].value_counts().sort_index()
print("Rating Difference Counts:")
print(rating_agreement)
```

```python
# Correlation between Self Rating and Manager Rating

rating_correlation = performance_df[
    ["SelfRating", "ManagerRating"]
].corr()
print("Rating Correlation:")
print(rating_correlation)
```

```python
# Check number of performance records per employee

performance_records = performance_df["EmployeeID"].value_counts()
print("Total performance records:", len(performance_df))
print("Unique employees in performance data:", performance_df["EmployeeID"].nunique())
print("\nPerformance records per employee:")
print(performance_records.describe())
```

```python
# Create employee-level performance summary

performance_summary = (
    performance_df
    .groupby("EmployeeID")
    .agg({
        "EnvironmentSatisfaction": "mean",
        "JobSatisfaction": "mean",
        "RelationshipSatisfaction": "mean",
        "WorkLifeBalance": "mean",
        "SelfRating": "mean",
        "ManagerRating": "mean"
    })
    .reset_index()
)
print("Performance summary created successfully!")
print("Rows:", performance_summary.shape[0])
print("Columns:", performance_summary.shape[1])
```

```python
# Merge employee and performance data

hr_df = employees_df.merge(
    performance_summary,
    on="EmployeeID",
    how="left"
)
print("Merged HR dataset created successfully!")
print("Rows:", hr_df.shape[0])
print("Columns:", hr_df.shape[1])
```

```python
# Check missing performance records after merge

performance_columns = [
    "EnvironmentSatisfaction",
    "JobSatisfaction",
    "RelationshipSatisfaction",
    "WorkLifeBalance",
    "SelfRating",
    "ManagerRating"
]
print("Missing performance records:")
print(hr_df[performance_columns].isnull().sum())
```

```python
# Count employees without performance records

employees_without_performance = hr_df["SelfRating"].isnull().sum()
print(
    "Employees without performance records:",
    employees_without_performance
)
```

```python
# Create Overall Performance Score

hr_df["OverallPerformanceScore"] = (
    hr_df["SelfRating"] +
    hr_df["ManagerRating"]
) / 2
print("Overall Performance Score created successfully!")
display(
    hr_df[
        [
            "EmployeeID",
            "SelfRating",
            "ManagerRating",
            "OverallPerformanceScore"
        ]
    ].head(10)
)
```

```python
# Satisfaction and Performance Correlation

satisfaction_performance = hr_df[
    [
        "EnvironmentSatisfaction",
        "JobSatisfaction",
        "RelationshipSatisfaction",
        "WorkLifeBalance",
        "OverallPerformanceScore"
    ]
].corr()
print("Satisfaction vs Performance Correlation:")
display(satisfaction_performance)
```

```python
# Calculate Overall Attrition Rate

total_employees = len(hr_df)
employees_left = (hr_df["Attrition"].eq("Yes").sum())
attrition_rate = round((employees_left / total_employees) * 100,2)
print("Total Employees:", total_employees)
print("Employees Left:", employees_left)
print("Overall Attrition Rate:", attrition_rate, "%")
```

## Final HR KPI Summary

This section summarizes the major HR metrics identified during exploratory data analysis and prepares the project for business reporting and dashboard development.

```python
# FINAL HR KPI SUMMARY
total_employees = hr_df["EmployeeID"].nunique()
employees_left = (hr_df["Attrition"].eq("Yes").sum())

attrition_rate = round(employees_left / total_employees * 100,2)

average_salary = round(hr_df["Salary"].mean(),2)

average_age = round(hr_df["Age"].mean(),2)

average_job_satisfaction = round(hr_df["JobSatisfaction"].mean(),2)

average_work_life_balance = round(hr_df["WorkLifeBalance"].mean(),2)

average_performance = round(hr_df["OverallPerformanceScore"].mean(),2)

print(" FINAL HR KPI SUMMARY ")

print(f"Total Employees: {total_employees}")
print(f"Employees Left: {employees_left}")
print(f"Overall Attrition Rate: {attrition_rate}%")
print(f"Average Age: {average_age}")
print(f"Average Salary: {average_salary}")
print(f"Average Job Satisfaction: {average_job_satisfaction}")
print(f"Average Work-Life Balance: {average_work_life_balance}")
print(f"Average Performance Score: {average_performance}")
```

```python
# Department KPI Summary

department_kpi = (
    hr_df
    .groupby("Department")
    .agg(
        Employee_Count=("EmployeeID", "nunique"),
        Average_Salary=("Salary", "mean"),
        Attrition_Rate=("Attrition", lambda x: (x == "Yes").mean() * 100),
        Average_Job_Satisfaction=("JobSatisfaction", "mean"),
        Average_Performance=("OverallPerformanceScore", "mean")
    )
    .round(2)
    .sort_values("Attrition_Rate", ascending=False)
)
display(department_kpi)
```

```python
# Save Department KPI Summary

department_kpi.to_csv("department_kpi_summary.csv")
print("Department KPI summary saved successfully!")
```
