```python
# Import required libraries
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

```python
# Import MySQL Connector
import mysql.connector

print("MySQL connector imported successfully!")
```

```python
# Connect Python to MySQL
connection = mysql.connector.connect(
    host="localhost",
    user="root",
    password="1612",
    database="enterprise_hr_analytics",
    port=3306
)

if connection.is_connected():
    print("MySQL connection successful!")
    print("Connected database: enterprise_hr_analytics")
```

```python
# Load employees table from MySQL

query_employees = """
select *
from employees;
"""

employees_df = pd.read_sql(query_employees, connection)


print("Employees data loaded successfully!")
print("Rows:", employees_df.shape[0])
print("Columns:", employees_df.shape[1])
```

```python
# Preview the first 10 employee records
employees_df.head(10)
```

```python
# Check the dimensions of the employee dataset
employees_df.shape
```

```python
# Load performance ratings table from MySQL

query_performance = """
select *
from performance_ratings;
"""

performance_df = pd.read_sql(query_performance, connection)

print("Performance ratings data loaded successfully!")
print("Rows:", performance_df.shape[0])
print("Columns:", performance_df.shape[1])
```

```python
# Check performance dataset dimension

print("Shape:", performance_df.shape)
print("\nColumns:")
print(performance_df.columns.to_list())
```

```python
# Preview first 10 performance records
performance_df.head(10)
```

```python
# Confirm both dataste are available

print("Employees Datafram:")
print(employees_df.shape)

print("\nPerformance Dataframe:")
print(performance_df.shape)
```

```python
# Check missing values in employees dataset

employee_missing = employees_df.isnull().sum()
print(employee_missing)
```

```python
# check missing values inperformance dataset

performance_missing = performance_df.isnull().sum()
print(performance_missing)
```

```python
# Check duplicate records

print("Duplicate employee records:",employees_df.duplicated().sum())

print("Duplicate employee records:", performance_df.duplicated().sum())
```

```python
# check data type of employee columns
employees_df.info()
```

```python
# Check data types of performance columns
performance_df.info()
```

```python
# Check EmployeeID uniqueness

print("Total employee records:", len(employees_df))
print("Unique EmployeeIDS:", employees_df["EmployeeID"].nunique())
print("Duplicate EmployeeIDs:", employees_df["EmployeeID"].duplicated().sum())
```

```python
#  Check EmployeeIDs in performance data

employee_ids = set(employees_df["EmployeeID"])
performance_employee_ids = set(performance_df["EmployeeID"])

missing_employee_ids = performance_employee_ids - employee_ids

print("Performance EmployeeIDs not found in employees table:", len(missing_employee_ids))
```
